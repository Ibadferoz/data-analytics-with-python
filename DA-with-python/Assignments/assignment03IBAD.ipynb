3. Write a program that accepts a list from the user and displays the largest number in the list, without using the built-in max() function.

```python
user_input = input("Enter list elements separated by spaces: ")
numbers = [float(x) for x in user_input.split()]

if numbers:
    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    print("Largest number:", largest)
else:
    print("No numbers entered.")
```

4. Write a program that rotates the element of a list so that the element at the first index moves to the
second index, the element in the second index moves to the third index, etc., and the element in the last
index moves to the first index.

```python
def rotate_right(lst):
    if not lst:
        return lst
    return [lst[-1]] + lst[:-1]

numbers = [43, 24, 54, 67, 43, 32, 26, 55, 86]
rotated = rotate_right(numbers)
print("Original:", numbers)
print("Rotated: ", rotated)
```

5. Write a program that input a string and ask user to delete a given word from a string. 

```python
def delete_word_simple():
    text = input("Enter a string: ")
    word_to_remove = input("Enter the word to delete: ")
    result = text.replace(word_to_remove + " ", "").replace(word_to_remove, "")

    print("\nUpdated String:")
    print(result)

delete_word_simple()
```

6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It
should print the date in the form March 12, 2021.

```python
from datetime import datetime

def reformat_date():
    date_str = input("Enter a date (mm/dd/yyyy): ")
    date_obj = datetime.strptime(date_str, "%m/%d/%Y")
    formatted_date = date_obj.strftime("%B %d, %Y")
    
    print("Formatted Date:", formatted_date)

reformat_date()
```

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be
like this 

```python
matrix = [
    [2, 21, 3, 14],
    [5, 2, 3, 16],
    [8, 9, 10, 62]
]
for index, row in enumerate(matrix, start=1):
    row_sum = sum(row)
    print(f"Sum of row {index} = {row_sum}")
```

9. Write a program to add two matrices of size n x m. 

```python
matrix_A = [
    [6, 42, 33],
    [76, 6, 45],
    [23, 2, 43]
]

matrix_B = [
    [9, 12, 24],
    [78, 7, 66],
    [3, 56, 43]
]

rows = len(matrix_A)
cols = len(matrix_A[0])

result = [[0 for _ in range(cols)] for _ in range(rows)]

for i in range(rows):
    for j in range(cols):
        result[i][j] = matrix_A[i][j] + matrix_B[i][j]

print("Sum of the two matrices:")
for row in result:
    print(row)
```

10. Write a program to multiply two matrices

```python
matrix_A = [
    [9, 7],
    [4, 2],
    [8, 3]
]

matrix_B = [
    [3, 9, 7],
    [1, 2, 4]
]

rows_A = len(matrix_A)
cols_A = len(matrix_A[0])
cols_B = len(matrix_B[0])

result = [[0 for _ in range(cols_B)] for _ in range(rows_A)]
for i in range(rows_A):
    for j in range(cols_B):
        for k in range(cols_A):
            result[i][j] += matrix_A[i][k] * matrix_B[k][j]

print("Result of Matrix Multiplication:")
for row in result:
    print(row)
```
