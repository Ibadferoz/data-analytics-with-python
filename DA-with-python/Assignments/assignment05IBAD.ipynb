QUESTION 1: Working with Lists
1.	Create a list called numbers containing at least 8 integers.
2.	Using a for loop, do the following:
o	Print all numbers in the list.
o	Print the sum of all numbers.
o	Print only the even numbers. 

```python
numbers = [12, 7, 24, 31, 18, 5, 40, 9]

total_sum = 0
even_numbers = []

print("All numbers in the list:")
for num in numbers:
    print(num)  
    total_sum += num 
    if num % 2 == 0:  
        even_numbers.append(num)

print("-" * 25)

print("Sum of all numbers:", total_sum)

print("-" * 25)

print("Even numbers in the list:")
for even in even_numbers:
    print(even)
```

2.	Using loops:
o	Print all keys and their values.
o	Calculate the average of the marks.
o	Display whether the student Passed (average ≥ 50) or Failed.


```python
student_data = {
    "Umer": [85, 78, 92, 66, 74],
    "Ibad": [45, 38, 50, 42, 30]
}

for name, marks in student_data.items():
    print(f"Student: {name}")
    print(f"Marks: {marks}")
    

    total_marks = 0
    count = 0
    for mark in marks:
        total_marks += mark
        count += 1
    
    average = total_marks / count
    print(f"Average: {average:.2f}")
    
    
    if average >= 50:
        print("Status: Passed")
    else:
        print("Status: Failed")
        
    print("-" * 25)
```

QUESTION 3: List of Dictionaries
1.	Create a list called employees.
2.	Each employee must be stored as a dictionary containing:
o	name
o	department
o	salary
3.	Using loops:
o	Print details of all employees.
o	Find the employee with the highest salary.
o	Calculate the total salary expense.
Requirements:
•	Minimum 3 employees.
•	Use comparison logic.
•	Use loops to process data.


```python
employees = [
    {"name": "Zarar", "department": "IT", "salary": 85000},
    {"name": "Usama", "department": "Finance", "salary": 95000},
    {"name": "Hamza", "department": "Marketing", "salary": 72000},
    {"name": "Maryam", "department": "HR", "salary": 68000}
]

print("--- All Employee Details ---")
for emp in employees:
    print(f"Name: {emp['name']} | Department: {emp['department']} | Salary: PKR {emp['salary']:,}")

print("\n" + "-" * 35 + "\n")

highest_salary_employee = employees[0]  
total_salary_expense = 0


for emp in employees:
    
     total_salary_expense += emp["salary"]
    
     if emp["salary"] > highest_salary_employee["salary"]:
        highest_salary_employee = emp


print(f"Total Salary Expense: PKR {total_salary_expense:,}")
print(f"Highest Paid Employee: {highest_salary_employee['name']} ({highest_salary_employee['department']}) with a salary of PKR {highest_salary_employee['salary']:,}")
```

QUESTION 4: While Loop with User Input
1.	Ask the user to enter numbers.
2.	Store the numbers in a list.
3.	Stop when the user enters -1.
4.	After the loop ends:
o	Print the complete list.
o	Print the largest number.
o	Print the smallest number.
Requirements:
•	Must use a while loop.
•	Must store input in a list.
•	Handle user input correctly.


```python
numbers = []

while True:
    user_input = float(input("Enter a number (or -1 to stop): "))
    

    if user_input == -1:
        break
    numbers.append(user_input)


if numbers:
    print("\n--- Results ---")
    print("Complete list:", numbers)
    print("Largest number:", max(numbers))
    print("Smallest number:", min(numbers))
else:
    print("\nNo numbers were entered before stopping.")
```

QUESTION 5: Word Frequency Counter
1.	Ask the user to enter a sentence.
2.	Convert the sentence into a list of words.
3.	Use a dictionary to count how many times each word appears.
4.	Print the word frequency clearly.
Example Format (for understanding only):
apple: 3
banana: 2
orange: 1
Requirements:
•	Use string splitting.
•	Use a dictionary for counting.
•	Use loops for processing


```python
sentence = input("Enter a sentence: ")

words = sentence.lower().split()

word_counts = {}

for word in words:

    word = word.strip(".,!?:;\"'")

    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print("\nWord Frequency:")
for word, count in word_counts.items():
    print(f"{word}: {count}")
```
